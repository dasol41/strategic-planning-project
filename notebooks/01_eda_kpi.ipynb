# Strategic Planning & Insights (Smart Appliances, U.S.)

This notebook:
- Loads the raw Statista export
- Transforms wide-year columns to a long KPI table
- Produces a few executive-ready charts


```
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

RAW_PATH = Path("../data/raw/mi_consumer_smart-home_united-states_usd_en_with_kmis.csv")  # rename your file or update this path
```

```
# Load
df = pd.read_csv(RAW_PATH)
df.shape, df.columns[:10]
```

```
# Transform (reuse the project function)
import sys
sys.path.append("../")
from src.load_transform import transform_wide_to_long

long_df = transform_wide_to_long(df)
long_df.head()
```

```
# Save processed
out_path = Path("../data/processed/metrics_long.csv")
out_path.parent.mkdir(parents=True, exist_ok=True)
long_df.to_csv(out_path, index=False)
out_path
```

```
# KPI snapshot
from src.kpi import compute_kpis
kpis = compute_kpis(long_df)
kpis
```

```
# Revenue trend (Total)
rev = long_df[(long_df["chart"]=="Revenue") & (long_df["name"]=="Total")].sort_values("year")
rev = rev[rev["value"].notna()]
plt.figure()
plt.plot(rev["year"], rev["value"])
plt.title("Revenue (Total)")
plt.xlabel("Year")
plt.ylabel(rev["unit"].iloc[0] if not rev.empty else "")
plt.show()
```

```
# Household Penetration trend (Smart Appliances)
pen = long_df[(long_df["chart"]=="Household Penetration Rate") & (long_df["name"]=="Smart Appliances")].sort_values("year")
pen = pen[pen["value"].notna()]
plt.figure()
plt.plot(pen["year"], pen["value"])
plt.title("Household Penetration Rate (Smart Appliances)")
plt.xlabel("Year")
plt.ylabel("percent")
plt.show()
```
